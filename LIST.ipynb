```python
#The core difference between mutable and immutable objects in Python comes down to whether you can change the object's value in memory after it has been created.
#• Mutable objects can be modified or changed in place without changing their identity (id()).
#• Immutable objects cannot be modified after they are created. Any operation that alters their value will result in the creation of a brand-new object in memory with a different identity.
```

```python
B = [4, 2, -1, 7, 6]

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False

    return True

count = 0

for i in B:
    if is_prime(i):
        count = count + 1

print(count)
```

```python
dates = ['1/1/2004','21/10/2009','15/5/2024','8/8/2019','22/2/2001']
#1.Create a list of leap yr dates only (filter)
leap_dates = []
for i in dates:
    year = int(i.split('/')[-1])
    if year%4==0:
        leap_dates.append(i)
print(leap_dates)
```

```python
#2count no of dates which either belong to summer or winter
count = 0
for i in dates:
    month = int(i.split('/')[1])
    if month in [2, 3, 4, 5, 10, 11, 12, 1]:
        count = count + 1

print(count)
```

```python
#3 create a list of dates with mdofies format 1/1/2004 to 1-1-2004
new_dates = []
for i in dates:
    new_dates.append(i.replace('/','-'))
print(new_dates)
```
