```python
def is_armstrong(number):
    original = number
    t = 0
    l=len(str(number))
    while number != 0:
        d = number % 10
        t = t + d**l
        number = number // 10
    
    if t == original:
        return True
    else:
        return False
```

```python
is_armstrong(1634)
```

```python
def is_prime(number):
    is_divisible = False
    for i in range(2,int(number)):
        if number%i==0:
            is_divisible = True
            break
    if not is_divisible:
        return True
    else:
        return False
```

```python
is_prime(6)
```

```python
#count of all prime no.
count = 0
for i in range(5000,10000):
    if is_prime(i):
        count = count + 1
print(count)
```

```python
#addition of all prime no.
total = 0
for i in range(5000,10000):
    if is_prime(i):
        total = total + i
print(total)
```

```python
#checked no. is palindrome or not using def
def is_palindrome(n):
    y=0
    a=n
    while n!=0:
        d=n%10
        n=n//10
        y=(y*10)+d
    if a==y:
        return True
    else:
        return False
```

```python
is_palindrome(2112)
```

```python
#use aboce to find sum of all palindrome no.
total = 0
for i in range(100,200):
    if is_palindrome:
        total = total + i
print(total)
```
